```python
# 1. Filter German summer medals without duplicates
ger_medals = df[
    df['Medal'].notnull() & 
    (df['Season'] == 'Summer') & 
    df['NOC'].isin(['GER', 'FRG'])
].drop_duplicates(['Games', 'Event', 'Medal'])

# 2. Count medals per year
medals_per_year = ger_medals['Year'].value_counts().sort_index().reset_index()
medals_per_year.columns = ['Year', 'Medals']

# 3. Mark host years (Berlin or Munich)
host_years = df[df['City'].isin(['Berlin', 'Munich']) & (df['Season'] == 'Summer')]['Year'].unique()
medals_per_year['is_host'] = medals_per_year['Year'].isin(host_years)

# 4. Plot using Seaborn
plt.figure(figsize=(14, 5))
sns.barplot(
    data=medals_per_year, 
    x='Year', 
    y='Medals', 
    hue='is_host', 
    palette={True: 'crimson', False: 'steelblue'},
    dodge=False
)

plt.xticks(rotation=90)
```

Q12. Which sport has the widest range of athlete ages, and which has the narrowest (most age-restrictive in practice)?**
consider sorting sports by IQR and showing only the extremes.*

```python
q75 = df.groupby('Sport')['Age'].quantile(0.75)
q25 = df.groupby('Sport')['Age'].quantile(0.25)
iqr = (q75 - q25).sort_values()

narrowest = iqr.head(5).index
widest = iqr.tail(5).index
selected_sports = narrowest.union(widest)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=df[df['Sport'].isin(selected_sports)],
    x='Sport',
    y='Age',
    order=iqr.loc[selected_sports].sort_values(ascending=False).index)
plt.xticks(rotation=90)
```
